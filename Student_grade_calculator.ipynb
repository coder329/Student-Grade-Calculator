# Student Grade Calculator

## Task 2 - Level 1 - Day 2

This project calculates student grades based on predefined grading conditions.

### Objectives
- Practice `if`, `elif`, and `else` statements
- Define clear marks ranges for each grade
- Handle invalid marks
- Display student marks and assigned grades

## Grading Criteria

| Marks | Grade |
|-------|-------|
| 90 - 100 | A |
| 80 - 89 | B |
| 70 - 79 | C |
| 60 - 69 | D |
| 0 - 59 | F |
| Below 0 or above 100 | Invalid |

```python
def calculate_grade(marks):
    if marks < 0 or marks > 100:
        return "Invalid"
    elif marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "F"
```

```python
students = {
    "Student 1": 95,
    "Student 2": 88,
    "Student 3": 76,
    "Student 4": 64,
    "Student 5": 52,
    "Student 6": 91,
    "Student 7": 83,
    "Student 8": 71,
    "Student 9": 59,
    "Student 10": 105
}

print("===== Student Grade Calculator =====")
print(f"{'Student':<15}{'Marks':<10}{'Grade'}")
print("-" * 35)

for student, marks in students.items():
    grade = calculate_grade(marks)
    print(f"{student:<15}{marks:<10}{grade}")
```

## Explanation

The program uses conditional statements:

- `if` checks whether marks are invalid.
- `elif` checks each grade range.
- `else` assigns grade F to marks below 60.

The program also demonstrates how conditional statements can be applied while processing multiple student records.

```python
test_marks = [95, 84, 73, 65, 45, 105, -5]

for marks in test_marks:
    print(f"Marks: {marks} -> Grade: {calculate_grade(marks)}")
```
